### Python Q01. Reverse a string

**Prompt (interviewer's comment):**
```python
# Write a function that reverses a string without using `[::-1]`.
```

**Question / given input:**
```python
text = "interview"
# Expected output: weivretni
```

**Solution:**

````python
def reverse_string(text):
    result = ""
    for char in text:
        result = char + result
    return result

print(reverse_string("interview"))  # weivretni
````

### Python Q02. Check whether a string is a palindrome

**Prompt (interviewer's comment):**
```python
# Write a function that checks whether a word reads the same forward and backward. Ignore letter case.
```

**Question / given input:**
```python
word = "Level"
# Expected output: True
```

**Solution:**

````python
def is_palindrome(word):
    cleaned = word.lower()
    return cleaned == cleaned[::-1]

print(is_palindrome("Level"))  # True
````

```python
# Expected output: {'h': 1, 'e': 1, 'l': 3, 'o': 2, 'w': 1, 'r': 1, 'd': 1}
```

### Python Q04. Find the second-largest number

**Prompt (interviewer's comment):**
```python
# Find the second-largest distinct number in a list without sorting the list.
```

**Question / given input:**
```python
numbers = [10, 5, 8, 10, 3, 8]
# Expected output: 8
```

**Solution:**

````python
def second_largest(numbers):
    largest = second = float("-inf")
    for number in numbers:
        if number > largest:
            second = largest
            largest = number
        elif largest > number > second:
            second = number
    if second == float("-inf"):
        return None
    return second

print(second_largest([10, 5, 8, 10, 3, 8]))  # 8
````

### Python Q05. Find duplicate elements

**Prompt (interviewer's comment):**
```python
# Return the duplicate values from a list, without repeating a duplicate in the result.
```

**Question / given input:**
```python
numbers = [1, 2, 3, 2, 4, 1, 5]
# Expected output: [2, 1]
```

**Solution:**

````python
def find_duplicates(numbers):
    seen = set()
    duplicates = []
    added = set()
    for number in numbers:
        if number in seen and number not in added:
            duplicates.append(number)
            added.add(number)
        seen.add(number)
    return duplicates

print(find_duplicates([1, 2, 3, 2, 4, 1, 5]))  # [2, 1]
````

### Python Q06. Check whether two strings are anagrams

**Prompt (interviewer's comment):**


**Question / given input:**
```python
first = "Dormitory"
second = "Dirty room"
# Expected output: True
```

**Solution:**

````python
def are_anagrams(first, second):
    a = sorted(first.replace(" ", "").lower())
    b = sorted(second.replace(" ", "").lower())
    return a == b

print(are_anagrams("Dormitory", "Dirty room"))  # True
````

### Python Q07. Generate the Fibonacci sequence

**Prompt (interviewer's comment):**
```python
# Generate the first n Fibonacci numbers, starting with 0 and 1.
```

**Question / given input:**
```python
n = 7
# Expected output: [0, 1, 1, 2, 3, 5, 8]
```

**Solution:**

````python
def fibonacci(n):
    sequence = []
    a, b = 0, 1
    for _ in range(n):
        sequence.append(a)
        a, b = b, a + b
    return sequence

print(fibonacci(7))  # [0, 1, 1, 2, 3, 5, 8]
````

### Python Q08. Calculate factorial

**Prompt (interviewer's comment):**
```python
# Write an iterative function to calculate the factorial of a non-negative integer.
```

**Question / given input:**
```python
n = 5
# Expected output: 120
```

**Solution:**

````python
def factorial(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers")
    result = 1
    for number in range(2, n + 1):
        result *= number
    return result

print(factorial(5))  # 120
````

### Python Q09. Check whether a number is prime

**Prompt (interviewer's comment):**
```python
# Return True if a number is prime; otherwise return False. Use an efficient loop.
```

**Question / given input:**
```python
number = 29
# Expected output: True
```

**Solution:**

````python
def is_prime(number):
    if number < 2:
        return False
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            return False
    return True

print(is_prime(29))  # True
````

### Python Q10. Find the missing number

**Prompt (interviewer's comment):**
```python
# A list contains distinct numbers from 1 to n with one number missing. Find the missing number.
```

**Question / given input:**
```python
numbers = [1, 2, 4, 5, 6]
# Expected output: 3
```

**Solution:**

````python
def missing_number(numbers, n):
    expected_sum = n * (n + 1) // 2
    return expected_sum - sum(numbers)

print(missing_number([1, 2, 4, 5, 6], 6))  # 3
````

### Python Q11. Solve the Two Sum problem

**Prompt (interviewer's comment):**
```python
# Return the indices of two numbers whose sum equals the target. Assume exactly one solution exists.
```

**Question / given input:**
```python
numbers = [2, 7, 11, 15]
target = 9
# Expected output: [0, 1]
```

**Solution:**

````python
def two_sum(numbers, target):
    seen = {}
    for index, number in enumerate(numbers):
        complement = target - number
        if complement in seen:
            return [seen[complement], index]
        seen[number] = index
    return []

print(two_sum([2, 7, 11, 15], 9))  # [0, 1]
````

### Python Q12. Flatten a nested list (one level)

**Prompt (interviewer's comment):**
```python
# Convert a list of lists into one flat list using a loop or comprehension.
```

**Question / given input:**
```python
nested = [[1, 2], [3, 4], [5]]
# Expected output: [1, 2, 3, 4, 5]
```

**Solution:**

````python
def flatten_list(nested):
    flat = []
    for sublist in nested:
        for item in sublist:
            flat.append(item)
    return flat

print(flatten_list([[1, 2], [3, 4], [5]]))
````

### Python Q13. Remove duplicates while preserving order


```python
items = [3, 1, 3, 2, 1, 4]

```

**Solution:**

````python
def remove_duplicates(items):
    result = []
    seen = set()
    for item in items:
        if item not in seen:
            result.append(item)
            seen.add(item)
    return result

print(remove_duplicates([3, 1, 3, 2, 1, 4]))
````

### Python Q15. Find the most frequent element

**Prompt (interviewer's comment):**
```python
# Return the element that occurs most often in a list. If there is a tie, return the first one encountered.
```

**Question / given input:**
```python
items = [4, 1, 4, 2, 1, 4, 1]
# Expected output: 4
```

**Solution:**

````python
def most_frequent(items):
    counts = {}
    for item in items:
        counts[item] = counts.get(item, 0) + 1
    return max(counts, key=counts.get)

print(most_frequent([4, 1, 4, 2, 1, 4, 1]))  # 4
````

### Python Q16. Sort a dictionary by its values

**Prompt (interviewer's comment):**
```python
# Sort a dictionary in ascending order based on its values.
```

**Question / given input:**
```python
marks = {"Asha": 82, "Ravi": 91, "Meera": 75}
# Expected output: {'Meera': 75, 'Asha': 82, 'Ravi': 91}
```

**Solution:**

````python
def sort_by_value(data):
    return dict(sorted(data.items(), key=lambda item: item[1]))

marks = {"Asha": 82, "Ravi": 91, "Meera": 75}
print(sort_by_value(marks))
````

### Python Q17. Merge two dictionaries

```python
first = {"a": 1, "b": 2}
second = {"b": 20, "c": 3}
# Expected output: {'a': 1, 'b': 20, 'c': 3}
```

### Python Q19. Handle division by zero

**Prompt (interviewer's comment):**


**Question / given input:**
```python
print(safe_divide(10, 0))
# Expected output: Cannot divide by zero
```

**Solution:**

````python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Cannot divide by zero"

print(safe_divide(10, 0))
````

### Python Q20. Count lines and words in text


**Question / given input:**
```python

# Expected output: Lines: 2, Words: 7
```

**Solution:**

# Part 2 — SQL (20 Questions)

The SQL exercises use **SQLite-compatible SQL**. Run the setup cell below once to create sample tables and insert data. Then attempt each question before running its solution.

````python
import sqlite3
import pandas as pd
from IPython.display import display

connection = sqlite3.connect(":memory:")
cursor = connection.cursor()

cursor.executescript("""
CREATE TABLE departments (
    department_id INTEGER PRIMARY KEY,
    department_name TEXT NOT NULL
);

CREATE TABLE employees (
    employee_id INTEGER PRIMARY KEY,
    employee_name TEXT NOT NULL,
    department_id INTEGER,
    salary INTEGER,
    manager_id INTEGER,
    city TEXT,
    FOREIGN KEY (department_id) REFERENCES departments(department_id)
);

CREATE TABLE customers (
    customer_id INTEGER PRIMARY KEY,
    customer_name TEXT NOT NULL,
    city TEXT
);

CREATE TABLE orders (
    order_id INTEGER PRIMARY KEY,
    customer_id INTEGER,
    amount REAL,
    order_date TEXT,
    FOREIGN KEY (customer_id) REFERENCES customers(customer_id)
);

INSERT INTO departments VALUES
(1, 'Engineering'), (2, 'HR'), (3, 'Sales');

INSERT INTO employees VALUES
(1, 'Asha', 1, 70000, NULL, 'Bengaluru'),
(2, 'Ravi', 1, 60000, 1, 'Mysuru'),
(3, 'Meera', 2, 50000, NULL, 'Bengaluru'),
(4, 'Kiran', 3, 55000, NULL, 'Pune'),
(5, 'Nisha', 1, 80000, 1, 'Bengaluru'),
(6, 'Arjun', 3, 45000, 4, 'Mysuru'),
(7, 'Divya', 2, 50000, 3, 'Pune');

INSERT INTO customers VALUES
(1, 'Anil', 'Bengaluru'),
(2, 'Bhavna', 'Mysuru'),
(3, 'Charan', 'Pune'),
(4, 'Deepa', 'Bengaluru');

INSERT INTO orders VALUES
(101, 1, 1200, '2026-01-10'),
(102, 1, 800, '2026-01-15'),
(103, 2, 1500, '2026-02-01'),
(104, 3, 700, '2026-02-05'),
(105, 2, 500, '2026-02-10');
""")

def run_sql(query):
    display(pd.read_sql_query(query, connection))

print("Sample database created successfully.")
````
